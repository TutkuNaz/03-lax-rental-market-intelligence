# LAX Rental Market Intelligence — Market Analysis

```python
from pathlib import Path
import json
import pandas as pd

ROOT = Path('..').resolve()
monthly = pd.read_csv(ROOT / 'data/processed/rental_transactions_monthly.csv')
annual = pd.read_csv(ROOT / 'data/processed/rental_market_annual.csv')
metrics = json.loads((ROOT / 'reports/metrics.json').read_text())
```

```python
monthly.groupby(['month_number', 'month']).transactions.sum().reset_index().sort_values('month_number')
```

```python
annual[['company', 'annual_transactions', 'annual_market_share_pct', 'gross_revenue_per_transaction_usd']].sort_values('annual_transactions', ascending=False).round(2)
```

```python
volatility = (
    monthly.groupby('company').monthly_market_share_pct
    .agg(lambda values: values.std(ddof=0))
    .sort_values(ascending=False)
)
volatility.round(3)
```

Monthly share volatility uses the population standard deviation across all 12 months, matching the SQL implementation. Revenue per transaction remains a descriptive ratio, not price, margin, or profit.
