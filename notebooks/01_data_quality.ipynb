# LAX Rental Market Intelligence — Data Quality

The source is the official LAWA CY2024 on-airport rental-car statistics. The original PDF remains external; extracted source tables are not committed.

```python
from pathlib import Path
import json
import pandas as pd
ROOT = Path('..').resolve()
monthly = pd.read_csv(ROOT / 'data/processed/rental_transactions_monthly.csv')
annual = pd.read_csv(ROOT / 'data/processed/rental_market_annual.csv')
metrics = json.loads((ROOT / 'reports/metrics.json').read_text())
```

```python
pd.DataFrame({
    'company_month_rows': [len(monthly)], 'companies': [annual.company.nunique()],
    'missing_cells': [int(monthly.isna().sum().sum() + annual.isna().sum().sum())],
    'annual_transactions': [int(annual.annual_transactions.sum())]
})
```
